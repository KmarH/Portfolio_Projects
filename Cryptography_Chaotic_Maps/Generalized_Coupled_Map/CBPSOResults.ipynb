```python
import numpy as np
import matplotlib.pyplot as plt
import random
from math import exp
import time as t
import statistics as s

def log_map(v, r, q):
    if v <= r:
        return (((-q / (r * r)) * ((r - v) * (r - v))) + q)
    else:
        return (((-q / ((1 - r) * (1 - r))) * ((r - v) * (r - v))) + q)

def sim_map(params, iterations):
    mu, mu1, nu, nu1, alpha = params
    x0 = 0.03
    y0 = 0.01
    x = [x0]
    y = [y0]
    for i in range(1, iterations):
        xn = (log_map(x[i - 1], nu, mu) + (alpha * (y[i - 1] - x[i - 1]))) % 1
        yn = (log_map(y[i - 1], nu1, mu1) + (alpha * (x[i - 1] - y[i - 1]))) % 1
        x.append(xn)
        y.append(yn)
    # Note that simulation starts from the second point. This way, an attacker can work without initial conditions
    return np.column_stack((x[1:], y[1:])) 
    

def objective_function(params):
    iterations = 30 # Fewer points and the system cannot be identified, more points and many points congregate in the same place
    estimated_output = sim_map(params,iterations)
    error = 1 / iterations * np.sum((true_output - estimated_output) ** 2)
    return error

# Plotting is handled as a function
def plot_maps(true_output, estimated_output):
    plt.figure(figsize=(12, 5))
    # Original Map
    plt.subplot(1, 2, 1)
    plt.scatter(true_output[:, 0], true_output[:, 1], s=1, color='blue')
    plt.title('Original Map')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True, linestyle='--', alpha=0.6)

    # Approximate Map
    plt.subplot(1, 2, 2)
    plt.scatter(estimated_output[:, 0], estimated_output[:, 1], s=1, color='red')
    plt.title('Approximate Map')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()

# Chaotic initialization for particles
def generate_logistic_positions(num_particles, x0=0.6, mu=4.0):
    x = x0
    positions = []
    for _ in range(num_particles):
        x = mu * x * (1 - x)
        positions.append(x)
    return positions

# Particle class
# Nathan Rooy on Github has a basic PSO implementation, and I used his code as a template to generate the classes
class Particle:
    def __init__(self, bounds, logistic_positions, particle_id):
        global num_dimensions
        self.position_i = [
            bounds[i][0] + (bounds[i][1] - bounds[i][0]) * logistic_positions[particle_id] 
            for i in range(num_dimensions)
        ]
        self.velocity_i = [random.uniform(-1, 1) for _ in range(num_dimensions)] 
        self.pos_best_i = []  # Best individual position
        self.err_best_i = -1  # Best individual error
        self.err_i = -1  # Current error

    def evaluate(self, costFunc):
        self.err_i = costFunc(self.position_i)
        if self.err_i < self.err_best_i or self.err_best_i == -1:
            self.pos_best_i = list(self.position_i)
            self.err_best_i = self.err_i

    def update_velocity(self, pos_best_g, w):
        c1, c2 = 2,2 # Corresponding to what was used for CBPSO2
        for i in range(num_dimensions):
            r1, r2 = random.random(), random.random()
            vel_cognitive = c1 * r1 * (self.pos_best_i[i] - self.position_i[i])
            vel_social = c2 * r2 * (pos_best_g[i] - self.position_i[i])
            self.velocity_i[i] = w * self.velocity_i[i] + vel_cognitive + vel_social

    def update_position(self, bounds):
        for i in range(num_dimensions):
            self.position_i[i] += self.velocity_i[i]
            self.position_i[i] = np.clip(self.position_i[i], bounds[i][0], bounds[i][1])

    def chaotic_search(self, global_best, a, bounds): 
        mu = 4.0
        x = random.uniform(0.1, 0.9)
        for _ in range(100):
            x = mu * x * (1 - x)
        for i in range(num_dimensions):
            search_range = a * global_best[i]
            new_pos = global_best[i] - search_range + (2 * search_range) * x
            self.position_i[i] = np.clip(new_pos, bounds[i][0], bounds[i][1])

class CBPSO:
    def __init__(self, costFunc, bounds, num_particles, maxiter):
        global num_dimensions
        num_dimensions = len(bounds)
        self.err_best_g = float("inf")  # Global best error
        pos_best_g = []  # Global best position
        a = 0.1  # Chaotic search range factor


        logistic_positions = generate_logistic_positions(num_particles) # Chaotic initialization of each particle
        change_err = []
        change_iter = []
        
        # The swarm is initialized
        swarm = [Particle(bounds, logistic_positions, i) for i in range(num_particles)]

        for i in range(maxiter):
            w_max, w_min, p = 0.9, 0.2, 100
            w = (w_max - w_min) * exp(-p * (i / maxiter) ** 2) + w_min # inertia weight as an exponential function of iterations

            for particle in swarm:
                particle.evaluate(costFunc)
                if particle.err_i < self.err_best_g:
                    pos_best_g = list(particle.position_i)
                    self.err_best_g = particle.err_i

            for particle in swarm:
                particle.update_velocity(pos_best_g, w)
                particle.update_position(bounds)

            for particle in swarm:
                if self.err_best_g > 1:
                    particle.chaotic_search(pos_best_g, a, bounds)
                elif 10 ** -(i + 2) < self.err_best_g < 10 ** -i:
                    particle.chaotic_search(pos_best_g, a / 2, bounds)

            change_iter.append(i+1)
            change_err.append(self.err_best_g)
            

        self.best_position = [float(param) for param in pos_best_g]
        print('FINAL:')
        print('Optimized parameters:', self.best_position)
        print('Optimized MSE objective function:', self.err_best_g)
       # Plotting MSE vs iterations. The plot_maps function is not used because it is for other types of plots
        plt.figure(figsize=(12, 5))
        plt.subplot(1, 2, 1)
        plt.plot(change_iter, change_err, color='blue', marker='o', markersize=2, linewidth=1)
        plt.title('MSE vs Iterations')
        plt.xlabel('Iteration')
        plt.ylabel('MSE')
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.show()
    

#--- RUN ---#
print("CBPSO optimization is performed...")
true_params = [0.91, 0.85, 0.9, 0.333, 0.3314]
true_output = sim_map(true_params,30)
true_output_graph = sim_map(true_params,3000)
bounds = [(0.5, 1), (0.5, 1), (0.5, 1), (0.01, 1), (0.01, 0.5)]
initial_bounds = [(0.01,0.1),(0.01,0.1)]

cycles = 100
besti = 0
bestmse = 10
success_cond = 0.009
successes = 0
temp_avg = []
avg_err = []

for i in range(cycles):
    print(f"Cycle {i + 1}")
    print(f'Original x1 and x2 values: {true_params}')
    start_time = t.time()
    pso = CBPSO(objective_function, bounds, num_particles=1000, maxiter=250)
    end_time = t.time()
    net_time = float(end_time-start_time)
    avg_err.append(pso.err_best_g)
    temp_avg.append(net_time)
    # Even if fewer points are worked on for optimization, it is plotted with all possible points to compare in the best way
    estimated_output = sim_map(pso.best_position,3000)
    plot_maps(true_output_graph, estimated_output)
    if pso.err_best_g <= bestmse:
        bestmse = pso.err_best_g
        besti = i
    if bestmse <= success_cond:
        successes += 0


avg_final = round(s.mean(avg_err),4) #sum(avg_err) / len(avg_err)
stdev_final = round(s.stdev(avg_err),4)
print(f"\nAverage error in {cycles} iterations:")
print(avg_final)
print(f"Standard deviation of MSE:")
print(stdev_final)
print(f"\nAverage error in {cycles} iterations:")
print(avg_final)
print(f"Standard deviation of MSE:")
print(stdev_final)
print(f'The best error occurred in cycle {besti + 1}, and was equal to: {bestmse}. '
f'Furthermore, if a MSE magnitude of 10^-3 is taken as success, it is determined that there were {successes} successes')
print(f"The average time was {np.mean(temp_avg)} with standard deviation of {np.std(temp_avg)}")
```
